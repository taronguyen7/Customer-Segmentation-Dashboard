```python
# Create a snapshot date (one day after the most recent transaction)
snapshot_date = df["InvoiceDate"].max() + pd.Timedelta(days=1)

print(snapshot_date)
```

```python
# Create the RFM table
rfm = df.groupby("Customer ID").agg({
    "InvoiceDate": lambda x: (snapshot_date - x.max()).days,
    "Invoice": "nunique",
    "Revenue": "sum"
})

# Rename the columns
rfm.columns = ["Recency", "Frequency", "Monetary"]

# Display the first five rows
rfm.head()
```

```python
rfm.head()
```

```python
# Summary statistics
rfm.describe()
```

```python
import matplotlib.pyplot as plt

rfm.hist(figsize=(12, 4), bins=30)

plt.tight_layout()
plt.show()
```

```python
# Recency score (lower recency = better)
rfm["R_Score"] = pd.qcut(
    rfm["Recency"],
    q=5,
    labels=[5, 4, 3, 2, 1]
).astype(int)

# Frequency score (higher frequency = better)
rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    q=5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

# Monetary score (higher spending = better)
rfm["M_Score"] = pd.qcut(
    rfm["Monetary"],
    q=5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

# Combine into one RFM score
rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str)
    + rfm["F_Score"].astype(str)
    + rfm["M_Score"].astype(str)
)

rfm.head()
```

```python
# Function to assign customer segments
def customer_segment(row):
    if row["R_Score"] >= 4 and row["F_Score"] >= 4 and row["M_Score"] >= 4:
        return "Champions"
    elif row["R_Score"] >= 3 and row["F_Score"] >= 4:
        return "Loyal Customers"
    elif row["R_Score"] >= 4 and row["F_Score"] >= 2:
        return "Potential Loyalists"
    elif row["R_Score"] <= 2 and row["F_Score"] >= 3:
        return "At Risk"
    else:
        return "Hibernating"

# Apply the function
rfm["Segment"] = rfm.apply(customer_segment, axis=1)
```

```python
rfm["Segment"].value_counts()
```

```python
def customer_segment(row):
    if row["R_Score"] >= 4 and row["F_Score"] >= 4 and row["M_Score"] >= 4:
        return "High-Value Customers"
    elif row["R_Score"] >= 3 and row["F_Score"] >= 4:
        return "Loyal Customers"
    elif row["R_Score"] >= 4 and row["F_Score"] >= 2:
        return "Emerging Customers"
    elif row["R_Score"] <= 2 and row["F_Score"] >= 3:
        return "At Risk"
    else:
        return "Inactive Customers"

rfm["Segment"] = rfm.apply(customer_segment, axis=1)
```

```python
rfm["Segment"].value_counts()
```

```python
segment_order = [
    "High-Value Customers",
    "Loyal Customers",
    "Emerging Customers",
    "At Risk",
    "Inactive Customers"
]

segment_counts = (
    rfm["Segment"]
    .value_counts()
    .reindex(segment_order)
)

print(segment_counts)
```

```python
import matplotlib.pyplot as plt

segment_order = [
    "High-Value Customers",
    "Loyal Customers",
    "Emerging Customers",
    "At Risk",
    "Inactive Customers"
]

segment_counts = (
    rfm["Segment"]
    .value_counts()
    .reindex(segment_order)
)

plt.figure(figsize=(8,5))
plt.bar(segment_counts.index, segment_counts.values)

plt.title("Customer Segments")
plt.xlabel("Customer Segment")
plt.ylabel("Number of Customers")

plt.xticks(rotation=20)

plt.tight_layout()
plt.show()
```

Customer Segmentation Analysis

The RFM analysis grouped the 5,881 customers into five distinct customer segments based on their purchasing behaviour. As shown in the chart, Inactive Customers represent the largest segment, with 2,345 customers. This indicates that a significant proportion of customers have not purchased recently or have low purchasing activity, suggesting opportunities for re-engagement campaigns.

The second largest segment is High-Value Customers with 1,300 customers. These customers purchase frequently, have made recent purchases, and generate the highest levels of revenue. Retaining this group should be a priority through loyalty programmes, personalised offers, and exclusive promotions.

The At Risk segment contains 824 customers. These customers have previously demonstrated valuable purchasing behaviour but have not purchased recently, making them suitable targets for win-back campaigns.

The Emerging Customers segment consists of 714 customers who have recently engaged with the business and show potential to become loyal customers if nurtured through targeted marketing initiatives.

Finally, Loyal Customers account for 698 customers. Although this is the smallest segment, these customers purchase consistently and represent a stable source of revenue. Maintaining strong relationships with this group through continued engagement and rewards can help encourage long-term customer retention.

Overall, the customer segmentation demonstrates that different groups exhibit different purchasing behaviours. This enables the business to develop tailored marketing strategies for each segment, improving customer retention, increasing customer lifetime value, and allocating marketing resources more effectively.
